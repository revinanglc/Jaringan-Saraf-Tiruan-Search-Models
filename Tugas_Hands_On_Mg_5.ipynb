```python
from collections import deque

def bfs(graph, start_node, target_node):
    """BFS untuk mencari jalur tercepat dalam level game."""

    queue = deque([start_node])
    visited = {start_node}
    paths = {start_node: [start_node]}

    print(f"Memulai permainan dari: {start_node}")

    while queue:
        current = queue.popleft()
        print(f"Memasuki {current}")

        # Jika sampai Boss
        if current == target_node:
            print("\n🎉 Boss ditemukan!")
            return paths[current]

        # Jelajahi level berikutnya
        for neighbor in graph.get(current, []):
            if neighbor not in visited:
                visited.add(neighbor)
                queue.append(neighbor)
                paths[neighbor] = paths[current] + [neighbor]
                print(f"  ➜ Membuka {neighbor} | Jalur: {paths[neighbor]}")

    print("\nBoss tidak dapat dicapai.")
    return None


# ============================================
# Graf Level Game
# ============================================
game_levels = {
    "Start": ["Level1", "Level2"],
    "Level1": ["Level3"],
    "Level2": ["Level4"],
    "Level3": ["Boss"],
    "Level4": ["Boss"],
    "Boss": []
}

print("=== Mencari Jalur Tercepat ke Boss ===")
path = bfs(game_levels, "Start", "Boss")

if path:
    print(f"\n🏆 Jalur tercepat ke Boss adalah: {path}")
```
